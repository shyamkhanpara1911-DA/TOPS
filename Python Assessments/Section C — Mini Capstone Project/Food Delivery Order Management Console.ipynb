```python
import json
import os

# ------------------ Order Class ------------------
class Order:
    def __init__(self, order_id, customer_name, items, total_amount, status="Pending"):
        self.order_id = order_id
        self.customer_name = customer_name
        self.items = items
        self.total_amount = total_amount
        self.status = status

    def to_dict(self):
        return {
            "order_id": self.order_id,
            "customer_name": self.customer_name,
            "items": self.items,
            "total_amount": self.total_amount,
            "status": self.status
        }


# ------------------ File Handling ------------------
FILENAME = "orders.json"

def load_orders():
    if not os.path.exists(FILENAME):
        return []
    try:
        with open(FILENAME, "r") as file:
            data = json.load(file)
            return [Order(**order) for order in data]
    except Exception:
        print("⚠️ Error reading file. Starting with empty data.")
        return []

def save_orders(orders):
    with open(FILENAME, "w") as file:
        json.dump([order.to_dict() for order in orders], file, indent=4)


# ------------------ Helper Functions ------------------
def generate_order_id(orders):
    return len(orders) + 1


def place_order(orders):
    print("\n--- Place New Order ---")

    # Validate name
    name = input("Enter customer name: ").strip()
    if not name:
        print("❌ Name cannot be empty.")
        return

    # Validate items
    items_input = input("Enter items (comma separated): ").strip()
    if not items_input:
        print("❌ Items cannot be empty.")
        return
    items = [item.strip() for item in items_input.split(",")]

    # Validate amount
    try:
        amount = float(input("Enter total amount: "))
    except ValueError:
        print("❌ Invalid amount. Please enter numeric value.")
        return

    order_id = generate_order_id(orders)
    new_order = Order(order_id, name, items, amount)

    orders.append(new_order)
    save_orders(orders)

    print(f"✅ Order {order_id} placed successfully!")


def view_orders(orders):
    print("\n--- All Orders ---")

    if not orders:
        print("No orders found.")
        return

    print(f"{'ID':<5} {'Name':<15} {'Items':<10} {'Amount':<10} {'Status':<12}")
    print("-" * 55)

    for order in orders:
        status_display = order.status
        if order.status == "Delivered":
            status_display = f"✔ {order.status}"  # highlight

        print(f"{order.order_id:<5} {order.customer_name:<15} {len(order.items):<10} {order.total_amount:<10} {status_display:<12}")


def search_order(orders):
    print("\n--- Search Order ---")
    try:
        oid = int(input("Enter Order ID: "))
    except ValueError:
        print("❌ Invalid ID.")
        return

    for order in orders:
        if order.order_id == oid:
            print("\nOrder Found:")
            print(f"ID: {order.order_id}")
            print(f"Name: {order.customer_name}")
            print(f"Items: {', '.join(order.items)}")
            print(f"Amount: {order.total_amount}")
            print(f"Status: {order.status}")

            # Option to mark delivered
            mark = input("Mark as Delivered? (y/n): ").lower()
            if mark == 'y':
                order.status = "Delivered"
                save_orders(orders)
                print("✅ Order marked as Delivered.")
            return

    print("❌ Order not found.")


# ------------------ Main Menu ------------------
def main():
    orders = load_orders()

    while True:
        print("\n===== Food Delivery System =====")
        print("1. Place New Order")
        print("2. View All Orders")
        print("3. Search Order by ID")
        print("4. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            place_order(orders)

        elif choice == "2":
            view_orders(orders)

        elif choice == "3":
            search_order(orders)

        elif choice == "4":
            print("👋 Exiting... Data saved.")
            break

        else:
            print("❌ Invalid choice. Try again.")


# Run Program
if __name__ == "__main__":
    main()
```
