```python
def Delivery_Fee_Calculator():
    try:
        order_value=int(input("Enter a order value (Rs) : "))
        delivery_distance=float(input("Enter a delivery distance (km) : "))
        if order_value <0 or delivery_distance < 0:
            raise ValueError("Order value and distance cannot be negative!")
        else: 
            if order_value > 500:
                print(f" delivery fee  in {delivery_distance}KM:0 RS ")
                print(f" order_value  :{order_value} RS")
                print(f" final amount :{order_value} RS")
            else:
                if delivery_distance < 5:
                    print(f" delivery fee in {delivery_distance} : 30 RS ")
                    print(f" order_value  :{order_value} RS")
                    print(f" final amount :{order_value + 30} RS")
                else:
                    print(f" delivery fee in {delivery_distance} : 60 RS ")
                    print(f" order_value  :{order_value} RS")
                    print(f" final amount :{order_value + 60} RS")    
    except ValueError:
        print("Error: invled data only num allow")        
    except Exception as e:
        print("somting went wrong",e)   
     

Delivery_Fee_Calculator()
                  
```
