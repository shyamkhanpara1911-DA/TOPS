```python
#task 2 
class Product:
    def __init__(self, price):
        self._price = price

    def get_price(self):
        return self._price

    def set_price(self, new_price):
        if new_price < 0:
            raise ValueError("Price cannot be negative")
        self._price = new_price


p = Product(500)

print(p.get_price())

p.set_price(800)
print(p.get_price())

# p.set_price(-100)  # ❌ error
```

```python
#task 3 
class Playlist:
    def __init__(self):
        self._songs = []   # private list

    def add_song(self, song):
        self._songs.append(song)

    def remove_song(self, song):
        if song in self._songs:
            self._songs.remove(song)

    def get_songs(self):
        return self._songs


pl = Playlist()

pl.add_song("Kesariya")
pl.add_song("Believer")

pl.remove_song("Kesariya")

print(pl.get_songs())
```

```python
#task 4 
from abc import ABC, abstractmethod

class PaymentMethod(ABC):
    
    @abstractmethod
    def pay(self, amount):
        pass


class UPI(PaymentMethod):
    def pay(self, amount):
        print(f"Paying ₹{amount} using UPI")


class CreditCard(PaymentMethod):
    def pay(self, amount):
        print(f"Paying ₹{amount} using Credit Card")


u = UPI()
c = CreditCard()

u.pay(500)
c.pay(1000)
```
