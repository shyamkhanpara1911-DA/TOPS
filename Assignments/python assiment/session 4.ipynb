```python
#task 2
def clean_brand_name(name):
    name = name.strip()            # remove spaces
    name = name.replace("-", " ")  # replace hyphen
    return name

print(clean_brand_name(" oneplus-Nord "))
```

```python
#task 3
text = "Apple iPhone 14 Pro Max"

words = text.split()

brand = words[0]
model = " ".join(words[1:])

print("Brand:", brand)
print("Model:", model)
```

```python
#task 4
def format_product_display(name, price):
    return f"{name} - ₹{price}"

print(format_product_display("Boat Earbuds", 1299))
```

```python
#task 5
products = [' mi-Band 5 ', ' SAMSUNG-Galaxy ', ' realme-Book ']

cleaned = []

for p in products:
    p = p.strip()                # remove spaces
    p = p.replace("-", " ")      # replace hyphen
    p = p.title()                # title case
    cleaned.append(p)

print(cleaned)
```
