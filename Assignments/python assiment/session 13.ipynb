```python
#task 1 
def reverse_string(s):
    if s == "":   # base condition
        return s
    return reverse_string(s[1:]) + s[0]

print(reverse_string("hello"))
```

```python
#task 2
def sum_playlist_durations(durations):
    if len(durations) == 0:  # base condition
        return 0
    return durations[0] + sum_playlist_durations(durations[1:])
print(sum_playlist_durations([120, 200, 150]))
```

```python
#task 3
count = 10

def update_count():
    count = 5
    print("Inside:", count)

update_count()
print("Outside:", count)
```

```python
#task 4
def count_likes(posts):
    total = 0
    
    for post in posts:
        total += post["likes"]
        

        if "replies" in post:
            total += count_likes(post["replies"])
    
    return total


data = [
    {"likes": 10, "replies": [
        {"likes": 5},
        {"likes": 3, "replies": [
            {"likes": 2}
        ]}
    ]},
    {"likes": 7}
]

print("Total Likes:", count_likes(data))
```

```python
#task 5
user_status = "Offline"   # global

def check_status():
    app_status = "Online"   # local
    
    print("Inside Function:")
    print("User Status:", user_status)
    print("App Status:", app_status)

print("Before Function:")
print("User Status:", user_status)

check_status()

print("After Function:")
print("User Status:", user_status)
```
